```csharp
// Интерфейс — контракт: любое животное, реализующее его,
// должно уметь издавать звук.
public interface IAnimal
{
    void MakeSound();
}

// Базовый класс
public class Animal : IAnimal
{
    // Статическое свойство принадлежит классу, а не отдельному животному.
    public static int TotalAnimals { get; private set; } = 0;

    // Свойства объекта
    public string Name { get; private set; }
    public string Species { get; protected set; }
    public int Age { get; private set; }

    // Скрытое поле — доступно только внутри Animal.
    private bool isHungry;

    // Конструктор по умолчанию
    public Animal() : this("Безымянное животное", "Неизвестный вид", 0)
    {
    }

    // Конструктор с параметрами
    public Animal(string name, string species, int age)
    {
        Name = name;
        Species = species;
        Age = age;
        isHungry = true;

        TotalAnimals++;
    }

    // Метод интерфейса
    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издаёт звук.");
    }

    // Обычный метод
    public void Eat()
    {
        isHungry = false;
        Console.WriteLine($"{Name} поело.");
    }

    public void ShowState()
    {
        string hungerState = isHungry ? "голодное" : "сытое";

        Console.WriteLine(
            $"{Name}: вид — {Species}, возраст — {Age}, состояние — {hungerState}."
        );
    }
}

// Пример наследника
public class Cat : Animal
{
    public Cat(string name, int age) : base(name, "Кошка", age)
    {
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Мяу!");
    }
}

// Проверка работы классов
Animal animal1 = new Animal("Барсик", "Кот", 3);
Cat cat1 = new Cat("Кот", 2);
Animal animal2 = new Animal();

animal1.MakeSound();
cat1.MakeSound();

animal1.ShowState();
animal1.Eat();
animal1.ShowState();

Console.WriteLine($"Всего создано животных: {Animal.TotalAnimals}");
```
