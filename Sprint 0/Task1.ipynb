```polyglot-notebook
class Habitat
{
    public string Name { get; set; }

    public Habitat(string name)
    {
        Name = name;
    }
}

class AnimalAggregation
{
    public string Name { get; set; }
    private Habitat habitat;

    public AnimalAggregation(string name, Habitat habitat)
    {
        Name = name;
        this.habitat = habitat;
    }

    public void ShowHabitat()
    {
        Console.WriteLine($"{Name} обитает: {habitat.Name}");
    }
}

Habitat forest = new Habitat("Лес");

AnimalAggregation fox = new AnimalAggregation("Лиса", forest);

fox.ShowHabitat();

Console.WriteLine($"Среда обитания отдельно: {forest.Name}");
```

```polyglot-notebook
class Heart
{
    public void Beat()
    {
        Console.WriteLine("Сердце бьется.");
    }
}

class AnimalComposition
{
    private Heart heart;

    public AnimalComposition()
    {
        heart = new Heart();
    }

    public void Live()
    {
        heart.Beat();
        Console.WriteLine("Животное живет..");
    }
}

AnimalComposition animalWithHeart = new AnimalComposition();

animalWithHeart.Live();
```

```polyglot-notebook
abstract class AnimalAbstract
{
    public abstract void MakeSound();

    public void Sleep()
    {
        Console.WriteLine("Животное спит.");
    }
}

class DogAbstract : AnimalAbstract
{
    public override void MakeSound()
    {
        Console.WriteLine("Гав!");
    }
}

AnimalAbstract myDogAbstract = new DogAbstract();

myDogAbstract.MakeSound();
myDogAbstract.Sleep();
```

```polyglot-notebook
class AnimalPoly
{
    public virtual void MakeSound()
    {
        Console.WriteLine("Животное издает звук");
    }
}

class DogPoly : AnimalPoly
{
    public override void MakeSound()
    {
        Console.WriteLine("Гав!");
    }
}

class CatPoly : AnimalPoly
{
    public override void MakeSound()
    {
        Console.WriteLine("Мяу!");
    }
}

AnimalPoly[] animals = { new DogPoly(), new CatPoly(), new AnimalPoly() };

foreach (AnimalPoly animal in animals)
{
    animal.MakeSound();
}
```

```polyglot-notebook
class Animal
{
    public string Name;

    public void Eat()
    {
        Console.WriteLine($"{Name} ест.");
    }
}

class Dog : Animal
{
    public void Bark()
    {
        Console.WriteLine($"{Name} говорит: Гав!");
    }
}

Dog myDog = new Dog();

myDog.Name = "Бобик";

myDog.Eat();
myDog.Bark();
```

```polyglot-notebook
class Animal
{
    private string name;
    private string species;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string Species
    {
        get { return species; }
        set { species = value; }
    }

    public void MakeSound()
    {
        Console.WriteLine("Животное издает звук");
    }
}

Animal myAnimal = new Animal();

myAnimal.Name = "Барсик";
myAnimal.Species = "Кот";

Console.WriteLine($"Имя: {myAnimal.Name}");
Console.WriteLine($"Вид: {myAnimal.Species}");

myAnimal.MakeSound();
```
