```polyglot-notebook
int target = 1;
bool contains = false;

foreach (int n in numbers)
{
    if (n == target)
    {
        contains = true;
        break;
    }
}

if (contains)
{
    Console.WriteLine($"Число {target} есть в массиве.");
}
else
{
    Console.WriteLine($"Числа {target} нет в массиве.");
}
```

```polyglot-notebook
int[] numbers = { 5, 8, 1, 3, 9, 6, 7, 2 };

int min = numbers[0];
int max = numbers[0];

foreach (int n in numbers)
{
    if (n < min) min = n;
    if (n > max) max = n;
}

Console.WriteLine($"Минимум: {min}");
Console.WriteLine($"Максимум: {max}");
```
