```csharp
public interface IAnimalFoodDIP
{
    void Feed();
}

public class AnimalFoodDIP : IAnimalFoodDIP
{
    public void Feed()
    {
        Console.WriteLine("Животное получает корм");
    }
}

public class AnimalDIP
{
    private IAnimalFoodDIP _food;

    public AnimalDIP(IAnimalFoodDIP food)
    {
        _food = food;
    }

    public void Eat()
    {
        _food.Feed();
    }
}

IAnimalFoodDIP foodDIP = new AnimalFoodDIP();
AnimalDIP animalDIP = new AnimalDIP(foodDIP);

animalDIP.Eat();
```

```csharp
public interface ICanEatISP
{
    void Eat();
}

public interface ICanWalkISP
{
    void Walk();
}

public interface ICanFlyISP
{
    void Fly();
}

public class DogISP : ICanEatISP, ICanWalkISP
{
    public void Eat()
    {
        Console.WriteLine("Собака ест");
    }

    public void Walk()
    {
        Console.WriteLine("Собака гуляет");
    }
}

public class BirdISP : ICanEatISP, ICanWalkISP, ICanFlyISP
{
    public void Eat()
    {
        Console.WriteLine("Птица ест");
    }

    public void Walk()
    {
        Console.WriteLine("Птица ходит");
    }

    public void Fly()
    {
        Console.WriteLine("Птица летает");
    }
}

DogISP dogISP = new DogISP();
BirdISP birdISP = new BirdISP();

dogISP.Eat();
dogISP.Walk();

birdISP.Eat();
birdISP.Fly();
```

```csharp
public abstract class AnimalLSP
{
    public string Name { get; set; }

    public AnimalLSP(string name)
    {
        Name = name;
    }

    public abstract void MakeSound();
}

public class DogLSP : AnimalLSP
{
    public DogLSP(string name) : base(name) { }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name}: Гав!");
    }
}

public class CatLSP : AnimalLSP
{
    public CatLSP(string name) : base(name) { }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name}: Мяу!");
    }
}

public void AnimalLSPInfo(AnimalLSP animal)
{
    animal.MakeSound();
}

AnimalLSP dogLSP = new DogLSP("Бобик");
AnimalLSP catLSP = new CatLSP("Барсик");

AnimalLSPInfo(dogLSP);
AnimalLSPInfo(catLSP);
```

```csharp
public abstract class AnimalOCP
{
    public string Name { get; set; }

    public AnimalOCP(string name)
    {
        Name = name;
    }

    public abstract void MakeSound();
}

public class DogOCP : AnimalOCP
{
    public DogOCP(string name) : base(name) { }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name}: Гав!");
    }
}

public class CatOCP : AnimalOCP
{
    public CatOCP(string name) : base(name) { }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name}: Мяу!");
    }
}

public void MakeAnimalSound(AnimalOCP animal)
{
    animal.MakeSound();
}

AnimalOCP dogOCP = new DogOCP("Бобик");
AnimalOCP catOCP = new CatOCP("Барсик");

MakeAnimalSound(dogOCP);
MakeAnimalSound(catOCP);
```

```csharp
public class AnimalSRP
{
    public string Name { get; set; }
    public string Species { get; set; }

    public AnimalSRP(string name, string species)
    {
        Name = name;
        Species = species;
    }

    public void MakeSound()
    {
        Console.WriteLine("Животное издает звук");
    }
}

public class AnimalPrinterSRP
{
    public void PrintInfo(AnimalSRP animal)
    {
        Console.WriteLine($"Имя: {animal.Name}, Вид: {animal.Species}");
    }
}

AnimalSRP animalSRP = new AnimalSRP("Барсик", "Кот");
AnimalPrinterSRP printerSRP = new AnimalPrinterSRP();

animalSRP.MakeSound();
printerSRP.PrintInfo(animalSRP);
```
