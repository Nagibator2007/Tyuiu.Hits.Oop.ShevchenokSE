```csharp
using System;
using System.Collections.Generic;

// Базовый класс Event
public class Event
{
    public int EventId { get; set; }
    public string EventName { get; set; }
    public DateTime Date { get; set; }

    protected List<string> Participants = new List<string>();

    public Event(int eventId, string eventName, DateTime date)
    {
        EventId = eventId;
        EventName = eventName;
        Date = date;
    }

    // Метод регистрации участника
    public virtual void RegisterParticipant(string participantName, string interest = "")
    {
        Participants.Add(participantName);
        Console.WriteLine($"{participantName} зарегистрирован на мероприятие {EventName}");
    }

    // Метод уведомления участников
    public virtual void NotifyParticipants()
    {
        Console.WriteLine($"Участники мероприятия {EventName} уведомлены.");
    }

    // Метод получения информации о мероприятии
    public virtual string GetEventDetails()
    {
        return $"ID: {EventId}, Название: {EventName}, Дата: {Date:dd.MM.yyyy}, " +
               $"Количество участников: {Participants.Count}";
    }
}

// Производный класс Conference
public class Conference : Event
{
    public int SpeakersCount { get; set; }

    public Conference(int eventId, string eventName, DateTime date, int speakersCount)
        : base(eventId, eventName, date)
    {
        SpeakersCount = speakersCount;
    }

    // Переопределение уведомления участников
    public override void NotifyParticipants()
    {
        Console.WriteLine($"Участники конференции {EventName} получили программу.");
        Console.WriteLine($"Количество спикеров: {SpeakersCount}");
    }

    public override string GetEventDetails()
    {
        return base.GetEventDetails() + $", Количество спикеров: {SpeakersCount}";
    }
}

// Производный класс Festival
public class Festival : Event
{
    public int BandsCount { get; set; }

    private Dictionary<string, string> ParticipantGenres =
        new Dictionary<string, string>();

    public Festival(int eventId, string eventName, DateTime date, int bandsCount)
        : base(eventId, eventName, date)
    {
        BandsCount = bandsCount;
    }

    // Переопределение регистрации с выбором музыкального жанра
    public override void RegisterParticipant(string participantName, string interest = "")
    {
        Participants.Add(participantName);

        string genre = string.IsNullOrWhiteSpace(interest)
            ? "Не выбран"
            : interest;

        ParticipantGenres[participantName] = genre;

        Console.WriteLine($"{participantName} зарегистрирован на фестиваль {EventName}. Жанр: {genre}");
    }

    public override string GetEventDetails()
    {
        return base.GetEventDetails() + $", Количество музыкальных групп: {BandsCount}";
    }
}

// Производный класс Seminar
public class Seminar : Event
{
    public List<string> Topics { get; set; }

    public Seminar(int eventId, string eventName, DateTime date, List<string> topics)
        : base(eventId, eventName, date)
    {
        Topics = topics;
    }

    // Переопределение получения информации о мероприятии
    public override string GetEventDetails()
    {
        string topics = string.Join(", ", Topics);

        return base.GetEventDetails() + $", Тематические области: {topics}";
    }
}

// Создание объектов
Conference conference = new Conference(
    1,
    "IT-конференция",
    new DateTime(2026, 12, 10),
    15
);

Festival festival = new Festival(
    2,
    "Музыкальный фестиваль",
    new DateTime(2026, 12, 15),
    8
);

Seminar seminar = new Seminar(
    3,
    "Семинар по программированию",
    new DateTime(2026, 12, 20),
    new List<string> { "C#", "ООП", "Разработка ПО" }
);

// Регистрация участников
conference.RegisterParticipant("Алексей");
festival.RegisterParticipant("Мария", "Рок");
seminar.RegisterParticipant("Иван");

// Демонстрация полиморфизма
Event[] events = { conference, festival, seminar };

foreach (Event currentEvent in events)
{
    Console.WriteLine();
    Console.WriteLine(currentEvent.GetEventDetails());
    currentEvent.NotifyParticipants();
}
```
